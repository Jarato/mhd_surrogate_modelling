# Comprehensive Latent Space Analysis

This notebook provides a complete analysis of a model's learned latent space dynamics. It loads the results from a single `latent_space_analysis.npz` file (generated by `gen_latent_prediction.py`) and visualizes:

1.  **Summary Metrics:** The overall MSE and R² score for the latent space rollout.
2.  **Rollout Error:** The MSE of the latent space prediction at each timestep.
3.  **Latent Trajectories:** A comparison of the predicted vs. ground truth evolution for individual latent dimensions.
4.  **Koopman Eigenvector Evolution:** The evolution of the latent state projected onto the learned Koopman eigenvectors, sorted by various criteria.
5.  **Koopman Eigenvalue Spectrum:** The distribution of the learned eigenvalues in the complex plane.
6.  **Koopman Mode Energy Spectrum:** The magnitude (energy) of each Koopman mode plotted against its temporal frequency.

```python
# Example path - update this to your specific run's output.
eval_file_path = Path("/cephfs/users/skowronek/Documents/PhD/nuclear_fusion_cooling/prediction/mhd_surrogate_modelling/experiments/study_4_2d_tckae/output/tc_size_exploration/ld512_M16_K16_Ktc16_gtc1.0_bwd_False/eval/latent_space_analysis/latent_space_analysis.npz")

# eval_file_path = Path("/cephfs/users/skowronek/Documents/PhD/nuclear_fusion_cooling/prediction/mhd_surrogate_modelling/experiments/study_4_2d_tckae/output/no_tc/ld256_M1_K64_Ktc0_gtc0.0_gi10.0_bwd_False/eval/latent_space_analysis/latent_space_analysis.npz")

if eval_file_path.exists():
    # --- Load Summary Metrics and Data for Colormaps ---
    with np.load(eval_file_path) as data:
        avg_mse = data['avg_rollout_mse']
        r_squared = data['r_squared_latent']
        # Load magnitudes to create a shared color scale for plots
        koopman_mode_magnitudes = data.get('koopman_mode_magnitudes')

    print("--- LATENT SPACE EVALUATION SUMMARY ---")
    print(f"Average Latent Space Rollout MSE: {avg_mse:.6e}")
    print(f"Latent Space R-squared (R²) Score: {r_squared:.4f}\n")

    # --- Define a Global Colormap Range ---
    # This ensures that colors represent the same energy level across different plots.
    if koopman_mode_magnitudes is not None:
        mag_min = koopman_mode_magnitudes.min()
        mag_max = koopman_mode_magnitudes.max()
        print(f"Global energy range for colormaps: [{mag_min:.2f}, {mag_max:.2f}]")
    else:
        mag_min, mag_max = None, None
        print("Could not find 'koopman_mode_magnitudes' for global colormap.")
else:
    print(f"ERROR: Evaluation file not found at {eval_file_path}")
    print("Please run gen_latent_prediction.py first and/or update the path.")
```
