# View Reconstruction Visualization Plots

This notebook finds and displays the reconstruction comparison plots (`.png` files) that were generated by the `visualize_reconstruction.py` script.

It serves as a centralized dashboard to visually inspect the reconstruction quality for a specific model and channel across different timesteps.

### 1. Setup and Imports

```python
from pathlib import Path
from IPython.display import display, Image, Markdown
```

### 2. Specify Run and Visualization Directory

**Action Required:** Change the `plots_dir` variable below to point to the specific directory containing the `.png` files you want to view. This directory is generated by the `visualize_reconstruction.py` script.

```python
# Example path - update this to your specific run's output.
plots_dir = Path("/raid/skowronek/eval/param_sweep_04_07_25/lr_1e-4_ld_1024_wr_1.0_wp_1.0_wl_10.0_we_0.1/visualize_reconstruction/plots")

# Find all the PNG files in the directory
if plots_dir.exists():
    plot_files = sorted(list(plots_dir.glob("*.png")))
    print(f"Found {len(plot_files)} plots in {plots_dir}")
else:
    plot_files = []
    print(f"Directory not found: {plots_dir}")
```

### 3. Display the Plots

```python
if not plot_files:
    print("No plots to display. Please run visualize_reconstruction.py first.")
else:
    for plot_file in plot_files:
        # Display the filename as a header
        display(Markdown(f'#### {plot_file.name}'))
        # Display the image
        display(Image(filename=plot_file, width=1200))
        display(Markdown("---"))
```
